### Ficha PEAS

| Elemento | Descripción |
|---|---|
| **Percepción (S)** | Tiempo de espera en minutos, nivel de urgencia y condición de cliente premium. |
| **Acciones (A)** | Mantener en nivel 1, escalar a nivel 2 o escalar a nivel 3. |
| **Entorno (E)** | Mesa de ayuda, cola de tickets, clientes, técnicos y acuerdos de nivel de servicio. |
| **Objetivo** | Atender primero los incidentes críticos y evitar esperas excesivas con un uso equilibrado del personal. |
| **Medida de desempeño (P)** | Cumplimiento del SLA, tiempo promedio de resolución, satisfacción del cliente y distribución adecuada de la carga. |

### Justificación de las reglas

La urgencia alta se deriva inmediatamente al nivel 3 porque puede representar una interrupción crítica. La urgencia media pasa al nivel 2 cuando la espera llega a 30 minutos o el cliente es premium; de lo contrario permanece en nivel 1. Una urgencia baja solo sube al nivel 2 después de 60 minutos, o desde 45 minutos si el cliente es premium. Así se protege la atención crítica, se respeta el SLA premium y se evita saturar innecesariamente los niveles especializados.

### Código

```python
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):
    # Decide el nivel de soporte y retorna una tupla (acción, motivo).
    if not isinstance(tiempo_espera_minutos, int) or tiempo_espera_minutos < 0:
        return "dato inválido", "el tiempo de espera debe ser un entero no negativo"
    if not isinstance(nivel_urgencia, str):
        return "dato inválido", "el nivel de urgencia debe ser texto"
    if not isinstance(cliente_premium, bool):
        return "dato inválido", "cliente_premium debe indicarse con True o False"

    urgencia = nivel_urgencia.strip().lower()
    if urgencia not in {"baja", "media", "alta"}:
        return "dato inválido", "la urgencia debe ser baja, media o alta"

    if urgencia == "alta":
        return "escalar a nivel 3", "urgencia alta: requiere atención especializada inmediata"

    if urgencia == "media":
        if cliente_premium:
            return "escalar a nivel 2", "urgencia media y cliente premium"
        if tiempo_espera_minutos >= 30:
            return "escalar a nivel 2", "urgencia media con 30 minutos o más de espera"
        return "mantener en nivel 1", "urgencia media con espera menor de 30 minutos"

    # Urgencia baja
    if cliente_premium and tiempo_espera_minutos >= 45:
        return "escalar a nivel 2", "cliente premium con 45 minutos o más de espera"
    if tiempo_espera_minutos >= 60:
        return "escalar a nivel 2", "ticket de urgencia baja con 60 minutos o más de espera"
    return "mantener en nivel 1", "urgencia baja y tiempo de espera dentro del límite"
```

### Simulación y pruebas

```python
casos_soporte = [
    # espera, urgencia, premium
    (5, "alta", False),
    (10, "media", False),
    (30, "media", False),  # caso límite de 30 minutos
    (5, "media", True),
    (44, "baja", True),
    (45, "baja", True),   # caso límite premium
    (59, "baja", False),
    (60, "baja", False),  # caso límite general
]

for espera, urgencia, premium in casos_soporte:
    accion, motivo = agente_soporte(espera, urgencia, premium)
    print(
        f"Espera={espera:>2} min | Urgencia={urgencia:<5} | Premium={premium} "
        f"-> {accion}: {motivo}"
    )
```
